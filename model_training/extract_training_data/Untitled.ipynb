```python


# Open the file in read-only mode to inspect its contents
import h5py
import pandas as pd

file_path = 'output/B_2020-06-09_002_example_training_data.h5'

# 1. Open the file and extract the data arrays
with h5py.File(file_path, 'r') as f:
    data = {
        'cell_label': f['cell_label'][:],
        'events': f['events'][:],
        'raw_indices': f['raw_indices'][:],
        'scores': f['scores'][:]
    }

print(data["events"][0])
```
